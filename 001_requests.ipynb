```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-opus-5"
```

```python
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": [{"type": "text", "text": text}]
    })

def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": [{"type": "text", "text": text}]
    })


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return message

message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in 1 sentence"
        }
    ]
)
for block in message.content:
    if getattr(block, "type", None) == "text":
        print(block.text)
        
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "Write another sentence"
        }
    ]
)
for block in message.content:
    if getattr(block, "type", None) == "text":
        print(block.text) 
```

```python
messages = []

add_user_message(messages, "Define quantum computing in 1 sentence")

answer = chat(messages)

def extract_text(message):
    return "".join(
        block.text for block in message.content
        if getattr(block, "type", None) == "text"
    )

add_assistant_message(messages, extract_text(answer))

add_user_message(messages, "Write another sentence")

answer = chat(messages)

for block in answer.content:
    if getattr(block, "type", None) == "text":
        print(block.text)
```
