```python
from dotenv import load_dotenv

load_dotenv()

from anthropic import Anthropic

client = Anthropic()
model = "claude-opus-5"
```

```python
def add_user_message(messages, text):
    messages.append({
        "role": "user",
        "content": [{"type": "text", "text": text}]
    })

def add_assistant_message(messages, text):
    messages.append({
        "role": "assistant",
        "content": [{"type": "text", "text": text}]
    })


def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
        }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params) 
    return message
```

```python
messages = []

add_user_message(messages, "Write a Python function that checks a string for duplicate characters.",
)
answer = chat(messages, system="Write a one sentence description of a fake database")

stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True,
)
print(stream)  # confirm it's created
for event in stream:
    print(event)
    try:
        print('dir:', [a for a in dir(event) if not a.startswith('_')])
        if hasattr(event, '__dict__'):
            print('event.__dict__ =', event.__dict__)
    except Exception as e:
        print('inspect error:', e)
    if getattr(event, "type", None) == "text":
        print(event.text)
```
